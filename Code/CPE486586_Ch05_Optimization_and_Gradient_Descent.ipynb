# 2. Automatic Differentiation in Pytorch to find Partial Derivative

```python
import torch

# Define input values with requires_grad=True to track gradients
w1 = torch.tensor(2.0, requires_grad=True)  # w_1 = w
x1 = torch.tensor(1.0, requires_grad=False) # x_1 = x
b1 = torch.tensor(1.0, requires_grad=False) # b_1 = b
y1 = torch.tensor(5.0, requires_grad=False) # y_1 = y

# Compute intermediate variables
w2 = w1 * x1
w3 = w2 + b1
w4 = torch.max(torch.tensor(0.0), w3)
w5 = y1 - w4
C = w5  # C = result

# Compute gradients
C.backward()

# Print the gradient of C with respect to w1
print("partial C/partial w1:", w1.grad.item())
```

## Write a python program to solve a constraint optimization  problem: Minimize/Maximize $f(x,y) = 81x^2 + y^2$ subject to $4x^2 + y^2  = 9$.

```python
import numpy as np
from scipy.optimize import minimize

# Define the objective function
def objective(x):
    x, y = x
    return 81 * x**2 + y**2

# Define the constraint function
def constraint(x):
    x, y = x
    return 4 * x**2 + y**2 - 9

# Initial guess for x and y
x0 = np.array([0.000, 0.01])

# Set up the optimization problem
constraints = {'type': 'eq', 'fun': constraint}
result = minimize(objective, x0, constraints=constraints)

# Print the results
print("Optimal values:")
print(f"x = {result.x[0]:.4f}")
print(f"y = {result.x[1]:.4f}")
print(f"Minimum value of f(x, y) = {result.fun:.4f}")
```

# 6. Minimization through Gradient Descent in Pytorch

$$
J(w_1) = (w_1 - 5)^2 + 10
$$

The objective is to find $w_1$ that minimizes $J(w_1)$.

```python
import matplotlib.pyplot as plt
import torch
plt.rcParams['font.family'] = 'Serif'
plt.rcParams['font.size'] = 15

# Define the cost function J(w_1)
def J(w_1):
  return (w_1 - 5)**2 + 10
```

### Initiailize $w_1$

```python
# Create a PyTorch tensor with requires_grad=True
w_1 = torch.tensor(0.0, requires_grad=True)
```

###  Define Optimizer: we are going to using Stochastic Gradient Descent

```python
import torch.optim as optim
optimizer = optim.SGD([w_1], lr=0.1)
```

### Perform the Gradient Descent to get $w_1$ that Minimizes the Cost Function

```python
w_1_values = []
losses = []
# Perform gradient descent
for i in range(100):
    optimizer.zero_grad()
    loss = J(w_1)
    loss.backward()
    optimizer.step()
    w_1_values.append(w_1.item())
    losses.append(loss.item())
    if (i + 1) % 10 == 0:
        print(f"Iteration {i + 1}: w_1 = {w_1.item():.4f}, Loss = {loss.item():.4f}")
```

### Visualize the Loss as a function of iteration (also called epochs)

```python
# Plot the value of w_1 as a function of iterations
plt.plot(losses)
plt.xlabel('Iteration')
plt.ylabel('loss')
plt.title('Loss as a function of iterations')
```

### Visualize $w_1$ as a function of iteration

```python
# Plot the value of w_1 as a function of iterations
plt.plot(w_1_values)
plt.xlabel('Iteration')
plt.ylabel('$w_1$')
plt.title('Value of $w_1$ as a function of iterations')
```

# 7. Constraint Optimization Using PyTorch

```python
# Define the objective function
import torch
def objective(x):
    x, y = x
    return 81 * x**2 + y**2

# Define the constraint function
def constraint(x):
    x, y = x
    return 4 * x**2 + y**2 - 9
```

### Penalty Method

```python
# Penalty method: Combine objective and constraint
def loss_fn(x, penalty_weight=10.0):
    obj = objective(x)
    constr_violation = constraint(x)**2  # Squared to ensure non-negativity
    return obj + penalty_weight * constr_violation
```

```python
# Initialize trainable parameters
x = torch.tensor([1.0, 1.0], requires_grad=True)  # Initial guess for x and y

# Set up the optimizer
optimizer = torch.optim.SGD([x], lr=0.0001)  # Learning rate can be adjusted
```

```python
# Perform gradient descent
for i in range(1000):
    # Zero the gradients
    optimizer.zero_grad()

    # Compute the loss
    loss = loss_fn(x)

    # Backpropagate to compute gradients
    loss.backward()

    # Update the parameters
    optimizer.step()

    # Print progress (optional)
    if (i + 1) % 50 == 0:
        print(f"Iteration {i + 1}: x = {x[0].item():.4f}, y = {x[1].item():.4f}, Loss = {loss.item():.4f}")

# Print the final results
print("\nOptimal values:")
print(f"x = {x[0].item():.4f}")
print(f"y = {x[1].item():.4f}")
print(f"Objective value = {objective(x).item():.4f}")
print(f"Constraint violation = {constraint(x).item():.4f}")
```

# 8. Constraints with Equality and Inequality

Minimize: $$f(x_1, x_2) = x_1^2 + 2x_2^2$$
Subject to: $$g(x_1, x_2) = x_1 + x_2 - 3 \leq 0$$

```python
class ConstrainedOptimization:
    """
    Minimize: f(x1, x2) = x1^2 + 2*x2^2
    Subject to: g(x1, x2) = x1 + x2 - 3 <= 0
    """
    def __init__(self):
        # Initialize variables (requires_grad=True for automatic differentiation)
        self.x = torch.tensor([5.0, 5.0], requires_grad=True)  # Start outside feasible region
        
        
    def objective(self, x):
        """Original objective function"""
        return x[0]**2 + 2*x[1]**2
    
    def constraint(self, x):
        """Inequality constraint g(x) <= 0"""
        return x[0] + x[1] - 3
    
    def penalty_method(self, penalty_weight=10.0, max_iter=1000, lr=0.01):
        """Solve using penalty method"""
        print(f"\n=== PENALTY METHOD (weight={penalty_weight}) ===")
        
        # Reset variables
        x = torch.tensor([5.0, 5.0], requires_grad=True)
        optimizer = optim.Adam([x], lr=lr)
        
        history = {'x': [], 'f': [], 'constraint': [], 'penalty': []}
        
        for i in range(max_iter):
            optimizer.zero_grad()
            
            # Objective function
            f = self.objective(x)
            
            # Constraint violation (max(0, g(x)) for inequality constraints)
            constraint_violation = torch.max(torch.tensor(0.0), self.constraint(x))
            
            # Penalty term
            penalty = penalty_weight * constraint_violation**2
            
            # Total loss = objective + penalty
            total_loss = f + penalty
            
            # Backpropagation
            total_loss.backward()
            optimizer.step()
            
            # Store history
            if i % 100 == 0:
                with torch.no_grad():
                    history['x'].append(x.clone().numpy())
                    history['f'].append(f.item())
                    history['constraint'].append(self.constraint(x).item())
                    history['penalty'].append(penalty.item())
                    
                print(f"Iter {i:3d}: x=({x[0].item():.4f}, {x[1].item():.4f}), "
                    f"f={f.item():.4f}, g(x)={self.constraint(x).item():.4f}, "
                    f"penalty={penalty.item():.4f}")
        
        return x.detach().numpy(), history
    
    
problem = ConstrainedOptimization()
solutions = {}

# Method 1: Penalty Method
sol_penalty, hist_penalty = problem.penalty_method(penalty_weight=100.0, max_iter=10000)
solutions['Penalty'] = (sol_penalty, hist_penalty)
```

# 9. Solving Partial Differential Equations using PyTorch

We want to solve a partial differential equation

$$
\cfrac{dy}{dx} = -2y
$$
with initial condition $y(0) = 1$.

The analytical solution is $y = e^{-2x}$.

We can guess the solution to be of the form polynomial $y(x) = c_0 + c_1 x + c_2 x^2 + \cdots$

Let's use degree of polynomial as $7$.

Here, out cost function is 

$$
J(x) = \cfrac{dy}{dx} + 2y
$$

and we would like to minimize $J^2(x)$.

```python
import torch
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Define the polynomial parameterization of y(x)
class PolynomialModel:
    def __init__(self, degree):
        # Initialize coefficients as trainable parameters
        self.coeffs = torch.randn(degree + 1, requires_grad=True)

    def forward(self, x):
        # Compute y(x) = c_0 + c_1*x + c_2*x^2 + ...
        y = torch.zeros_like(x)
        for i, coeff in enumerate(self.coeffs):
            y += coeff * (x ** i)
        return y

# Step 2: Define the loss function (residual of the ODE + initial condition penalty)
def ode_loss(model, x):
    y = model.forward(x)  # Predict y(x)
    dy_dx = torch.autograd.grad(y, x, grad_outputs=torch.ones_like(y), create_graph=True)[0]  # Compute dy/dx
    
    # Residual of the ODE: dy/dx + 2y
    residual = dy_dx + 2 * y
    
    # Penalty for initial condition y(0) = 1
    initial_condition_penalty = (model.forward(torch.tensor([[0.0]], requires_grad=True)) - 1.0)**2
    
    # Total loss: mean squared residual + initial condition penalty
    return torch.mean(residual**2) + initial_condition_penalty

# Step 3: Training setup
degree = 7  # Degree of the polynomial
model = PolynomialModel(degree)
optimizer = optim.Adam([model.coeffs], lr=0.001)

# Generate training data (x values)
x_data = torch.linspace(0, 2, 100).view(-1, 1)  # x ranges from 0 to 2
x_data.requires_grad_(True)  # Ensure x_data requires gradients for autograd

# Step 4: Training loop
num_epochs = 50000
for epoch in range(num_epochs):
    optimizer.zero_grad()       # Zero gradients
    loss = ode_loss(model, x_data)  # Compute loss
    loss.backward()             # Backpropagation
    optimizer.step()            # Update parameters
    
    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {loss.item()}")

# Step 5: Compare with analytical solution
x_test = torch.linspace(0, 2, 100).view(-1, 1)
y_pred = model.forward(x_test).detach().numpy()
y_true = np.exp(-2 * x_test.numpy())  # Analytical solution

# Plot results
plt.plot(x_test.numpy(), y_pred, label="Predicted Solution")
plt.plot(x_test.numpy(), y_true, label="Analytical Solution", linestyle="--")
plt.xlabel("x")
plt.ylabel("y(x)")
plt.legend()
plt.title("Solution of ODE using PyTorch Autograd (Polynomial Parameterization)")
plt.show()
```
