# 1. The Longest Substring

```python
def LongestSubstring(s):
    # Check if the string is empty
    if not s:
        print("empty string")
        return

    # Check if the string length exceeds 20 characters
    if len(s) > 20:
        print("maximum length exceeded")
        return

    max_substring = ""
    current_substring = ""

    # Iterate through the string to find the longest substring with the same characters
    for i in range(len(s)):
        if i == 0 or s[i] == s[i - 1]:
            current_substring += s[i]
        else:
            if len(current_substring) > len(max_substring):
                max_substring = current_substring
            current_substring = s[i]

    # Final check in case the longest substring is at the end of the string
    if len(current_substring) > len(max_substring):
        max_substring = current_substring

    # Print the result
    print(max_substring)

# Example usage
LongestSubstring("22427777232888823")  # Output: '7777'
LongestSubstring("")                  # Output: 'empty string'
LongestSubstring("a" * 21)            # Output: 'maximum length exceeded'
LongestSubstring("aaabbbbaacccdd")    # Output: 'bbbb'
```

# 3. Working with Data Types in Python

```python
# x should be a list or tuple input
def lenmaxmin(x):

    # Calculating the length, maximum, and minimum value of the list/tuple
    num_elements = len(x)
    maxi = max(x)
    mini = min(x)

    # This returns the value as tuple
    return (num_elements, maxi, mini)


print(lenmaxmin((12,6,4,65)))
print(lenmaxmin([311,45,9]))
print(lenmaxmin([311,45,9, 74, 10222, 89, -93]))
```

# 9. Solving System of Linear Equations

```python
import torch
A = torch.tensor([[4, 3, 2], [-2, 2, 3], [3, -5, 2]], dtype=torch.float32)
B = torch.tensor([[25], [-10], [-4]], dtype=torch.float32)
def linearsolve(A, B):
    """
    Args:
        A (torch.Tensor): System Matrix
        B (torch.Tensor): Right Hand Side
        
    Returns:
        Solution Vector
    """
    X = torch.linalg.solve(A, B)
    return X
    
X = linearsolve(A, B)
print(X)
```

# 10. Elementary Operations on Matrices

```python
def rowswap(matrix: torch.Tensor,
            source_idx: int,
            target_idx: int) -> torch.Tensor:
    """
    Swap two rows of a 2-D tensor **in-place**.

    Args:
        matrix: 2-D tensor whose rows are to be swapped.
        source_idx: Index of the first row.
        target_idx: Index of the row to swap with.

    Returns:
        The same tensor object with the requested rows exchanged.
    """
    matrix[[source_idx, target_idx]] = matrix[[target_idx, source_idx]]
    return matrix

    
```

```python
def rowscale(matrix: torch.Tensor,
                row_idx: int,
                scale_factor: float) -> torch.Tensor:
    """
    Multiply every element of a given row by a scalar **in-place**.

    Args:
        matrix: 2-D tensor.
        row_idx: Index of the row to scale.
        scale_factor: Value to multiply the row by.

    Returns:
        The same tensor object with the row scaled.
    """
    matrix[row_idx] *= scale_factor
    return matrix
```

```python
def rowreplacement(matrix: torch.Tensor,
                    target_idx: int,
                    source_idx: int,
                    target_scale: float,
                    source_scale: float) -> torch.Tensor:
    """
    Perform the elementary row operation
        R[target_idx] = target_scale * R[target_idx] + source_scale * R[source_idx]
    **in-place**.

    Args:
        matrix: 2-D tensor.
        target_idx: Index of the row to be overwritten.
        source_idx: Index of the row that contributes to the linear combination.
        target_scale: Coefficient for the target row.
        source_scale: Coefficient for the source row.

    Returns:
        The same tensor object after the row replacement.
    """
    
    matrix[target_idx].mul_(target_scale).add_(matrix[source_idx], alpha=source_scale)
    return matrix
    
```

```python
A = torch.arange(6, dtype=torch.float32).view(2, 3)
print("Original:\n", A)

rowswap(A, 0, 1)
print("After row-swap:\n", A)

rowscale(A, 0, 10)
print("After row-scale (row 0 *= 10):\n", A)

rowreplacement(A, 1, 0, 1, -1)
print("After row-replacement (R1 = R1 - R0):\n", A)
```

```python
from typing import Tuple
def to_row_echelon_form(matrix: torch.Tensor) -> torch.Tensor:
    """
    Convert a matrix to row echelon form (REF) using elementary row operations.
    
    In REF:
    - Each row has a leading 1 (pivot)
    - All entries below each pivot are 0
    - Each pivot is to the right of the pivot in the row above
    
    Args:
        matrix: 2-D tensor to convert to REF
        
    Returns:
        The same tensor object converted to row echelon form
    """
    if matrix.dim() != 2:
        raise ValueError("Matrix must be 2-dimensional")
    
    rows, cols = matrix.shape
    tolerance = 1e-05
    
    pivot_idx = 0
    for c in range(0, cols):
        # Exit if no more rows available for pivots
        if pivot_idx >= rows:
            break
        # find a non-zero element in the  current column if the first element is zero
        
        if(abs(matrix[pivot_idx, c]) <= tolerance):
            matrix[pivot_idx, c] = 0.0
            # find the maximum element in the current column from pivot_idx onwards
            max_value = matrix[pivot_idx:, c].max()  # Search only from pivot_idx onwards
            if abs(max_value) <= tolerance:  # No valid pivot found in this column
                continue  # Skip this column, don't increment pivot_idx
            max_index = matrix[pivot_idx:, c].argmax() + pivot_idx  # Adjust index
            # swap current pivot row with max row
            matrix = rowswap(matrix, pivot_idx, int(max_index))
        
        # Check if pivot is still zero after potential swap
        if abs(matrix[pivot_idx, c]) <= tolerance:
            continue  # Skip this column, don't increment pivot_idx
            
        if (matrix[pivot_idx,c] != 1):
            matrix = rowscale(matrix, pivot_idx, 1.0/matrix[pivot_idx,c])

        for r in range(pivot_idx+1, rows):
            if(abs(matrix[r, c]) <= tolerance):
                matrix[r, c] = 0.0
                continue
            matrix = rowreplacement(matrix, r, pivot_idx, 1.0, -1.0*matrix[r,c])
        pivot_idx = pivot_idx + 1  # Only increment when we successfully process a column
    
    return matrix
```
